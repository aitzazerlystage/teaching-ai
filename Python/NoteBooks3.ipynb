### Increment Operators

```python
x = 0
print("Starting value:", x)

x += 1   # same as x = x + 1
print("After increment:", x)

x += 5   # add 5 more
print("After adding 5:", x)
```

```python
y = 10
print("Starting value:", y)

y -= 1   # same as y = y - 1
print("After decrement:", y)

y -= 3   # subtract 3 more
print("After subtracting 3:", y)
```

### NESTED LOOPS & CONDITIONALS

```python
# Print a multiplication table
for i in range(1,4):       # Outer loop
    for j in range(1,10):   # Inner loop
        print(i * j) #2 x 1


    print("-----")  # separator after each row
```

```python
i = 1
while i <= 3: 
         # Outer loop
    j = 1

    while j <= 3:  # Inner loop
        print(i * j)
        j += 1
    print("-----")
    i += 1
```

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are an adult.")
    
    if age >= 60:
        print("You are also a senior citizen.")
    else:
        print("You are not a senior citizen.")

        
else:
    print("You are a minor.")
```

Break & Continue Statements

```python
# Stop the loop when number 3 is found
for i in range(1, 6):
    if i == 3:
        print("Found 3! Stopping the loop.")
        break
        
    print("Number:", i)
```

```python
# Skip number 3 but keep looping
for i in range(1, 6):
    if i == 3:
        print("Skipping 3")
        continue
    print("Number:", i)
    print("Number:", i)
    print("Number:", i)



```

```python
# Keep asking numbers, skip negative ones
while True:
    num = int(input("Enter a number (0 to quit): "))
    
    if num == 0:
        print("Goodbye!")
        break
    
    if num < 0:
        print("Negative numbers are skipped!")
        continue
    
    print("You entered:", num)
```

```python
# User has 3 attempts to enter the correct password
attempts = 3

while attempts > 0:
    password = input("Enter password: ")
    
    if password == "secret123":
        print("Access Granted ✅")
        break
    else:
        attempts -= 1
        print("Wrong password. Attempts left:", attempts)

if attempts == 0:
    print("Access Denied ❌ Too many wrong attempts.")
```
