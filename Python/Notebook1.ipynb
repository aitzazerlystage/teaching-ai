```python
# Enumerate
names = ["Alice", "Bob", "Charlie"]
for i, name in enumerate(names):
    print(f" index : {i}: {name}")

# Zip
scores = [85, 92, 78]
for name_in_this_loop, score_in_second_list in zip(names, scores):
    print(f"{name_in_this_loop} scored {score_in_second_list}")

# Continue and break
for i in range(10):
    if i % 2 == 0:
        continue
    if i > 7:
        break
    print(i)
```

```python
value = "100"

# Convert string to int
int_value = int(value)

# Convert int to float
float_value = float(int_value)

# Convert float to string
string_value = str(float_value)

print(int_value, float_value, string_value)
print(type(int_value), type(float_value), type(string_value))

# Type-safe addition
if isinstance(int_value, int):
    print("Safe to add:", int_value + 20)
```

```python
import math

angle = 45
radians = math.radians(angle)

print("Sine of 45°:", math.sin(radians))
print("Factorial of 5:", math.factorial(5))
print("Square root of 49:", math.sqrt(49))
print("Power of 2^10:", math.pow(2, 10))
```

```python
import datetime

# Get today's date
today = datetime.date.today()
print("Today's date:", today)

# Get current date and time
now = datetime.datetime.now()
print("Current date and time:", now)

# Create a specific date
my_birthday = datetime.date(2000, 5, 15)
print("My birthday is on:", my_birthday)
```

```python
import random

# Get a random number between 1 and 10
random_number = random.randint(1, 100)
print("Random number between 1 and 10:", random_number)

# Choose a random item from a list
fruits = ["apple", "banana", "cherry", "mango"]
random_fruit = random.choice(fruits)
print("Random fruit:", random_fruit)

# Get a random float between 0 and 1
random_float = random.random()
print("Random float between 0 and 1:", random_float)
```

```python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Accessing items
print(fruits[0])  # prints "apple"

# Adding items
fruits.append("mango")
print(fruits)

# Changing an item
fruits[1] = "orange"
print(fruits)

# Removing an item
fruits.remove("cherry")
print(fruits)

# Checking length
print("Number of fruits:", len(fruits))
```

```python
for fruit in fruits:
    print("I like", fruit)
```

```python
# Creating a dictionary
student = {
    "name": "John",
    "age": 20,
    "grade": "A"
   }

# Accessing values
print(student["name"])
print(student["age"])

# Adding a new key-value pair
student["email"] = "john@example.com"
print(student["email"])

# Updating a value
student["grade"] = "A+"

# Removing a key
del student["age"]

# Printing the updated dictionary
print(student)
```

```python
for i, j in student.items():
    print(i, ":", j)
```
