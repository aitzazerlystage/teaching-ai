Methods of List

```python
# Create a list of animals
animals = ["cat", "dog", "elephant"]
print("Original list:", animals)

# Add an item at the end
animals.append("lion")

# Insert an item at index 1
animals.insert(1, "tiger")

# Remove a specific item by value
animals.remove("dog")

# Remove the last item using pop()
animals.pop()

# Sort the list in ascending order
animals.sort()

# Reverse the order of the list
animals.reverse()

# Extend the list with more items
animals.extend(["giraffe", "zebra"])

# Get the index of a specific item
index = animals.index("elephant")

# Count how many times an item appears
count = animals.count("cat")

# Make a copy of the list
copy_animals = animals.copy()

# Clear the list
animals.clear()

# Display results
print("Index of elephant:", index)
print("Count of cat:", count)
print("Copied animal list:", copy_animals)
print("Cleared animal list:", animals)
```

Nested Tuples

```python
nested = ([1, 2], [3, 4])
nested[0][0] = 10
print(nested)  # Output: ([[10, 2], [3, 4]])
```

Dictionary

```python
my_dict = {"name": "Ali", "age": 20}
my_dict["age"] = 21            # Update value
my_dict["city"] = "Lahore"     # Add new key-value
del my_dict["name"]            # Delete a key

print(my_dict)  # Output: {'age': 21, 'city': 'Lahore'}
```

Methods of Dictionaries

```python
# Create a dictionary
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

# Add a new key-value pair
student["email"] = "alice@example.com"

# Update an existing value
student.update({"age": 21})

# Copy the dictionary
copy_student = student.copy()

# Clear all items from the original dictionary
student.clear()

# Print outputs
print("Copied student dictionary:", copy_student)
print("Original student dictionary after clear():", student)
```

Methods of Strings

```python
text = " Hello Python World! "

print(text.lower())
print(text.upper())
print(text.strip())
print(text.replace("Python", "Coding"))
print(text.split())
print(text.find("Python"))
print(text.startswith(" Hello"))
print(text.endswith("World! "))
print(text.count("o"))
print(text.title())
print(text.isdigit())
print(text.islower())
print(text.isupper())
```

Logical Operators

```python


is_sunny = True     
is_weekend = False  

# AND → You go for a picnic only if it is sunny AND it's the weekend
print(is_sunny and is_weekend)  # False → It’s sunny but not a weekend, so no picnic

# OR → You go outside if it is sunny OR it's the weekend
print((is_sunny or is_weekend) and (not is_weekend))   # True → It’s sunny, even if it’s not weekend

# NOT → You stay home if it is NOT sunny
print(not is_sunny)             # False → It IS sunny, so NOT sunny is False
```
